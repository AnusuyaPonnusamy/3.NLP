```python
import pandas as pd
import numpy as np
import itertools
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn import metrics
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('news_dataset.csv', index_col=None)
```

```python
df
```

```python
dataset=df
```

```python
y=dataset["label"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(dataset['text'], y, test_size=0.33, random_state=53)
```

```python
count_vectorizer=CountVectorizer(stop_words='english')  # stop_words like is,was,the,in etc will be removed
count_train=count_vectorizer.fit_transform(X_train) # fit method creates model. fit_transform creates model + predict
print(count_train)
count_test=count_vectorizer.transform(X_test) # convert the new text into the same numerical format that was already learned from the training data.
```

```python
dir(count_train)
```

```python
len(count_vectorizer.get_feature_names_out())
```

```python
print(count_train.toarray())
```

```python
# So far we have created the vectors for train and test dataset which are now available as numbers. Now we are going to create the model using
# multinomialnb.
```

# MULTINOMIALNB

```python
clf=MultinomialNB()
clf.fit(count_train, y_train)
pred=clf.predict(count_test)
score=metrics.accuracy_score(y_test, pred)
print("accuracy: %0.3f" %score)

cm=metrics.confusion_matrix(y_test, pred, labels=['fake','real'])
```

```python
from sklearn.metrics import classification_report
report=classification_report(y_test,pred)
```

```python
print(report)
```

# Logistic Regression

```python
clf_1=LogisticRegression(max_iter=1000,random_state=0)
clf_1.fit(count_train, y_train)
pred=clf_1.predict(count_test)
score=metrics.accuracy_score(y_test, pred)
print("accuracy: %0.3f" %score)

cm=metrics.confusion_matrix(y_test, pred, labels=['fake','real'])
```

# PassiveAggressiveClassifier

```python
clf_2=PassiveAggressiveClassifier()
clf_2.fit(count_train, y_train)
pred=clf_2.predict(count_test)
score=metrics.accuracy_score(y_test, pred)
print("accuracy: %0.3f" %score)

cm=metrics.confusion_matrix(y_test, pred, labels=['fake','real'])
```

# RandomForest

```python
clf_3=RandomForestClassifier(n_estimators=10, criterion='entropy', random_state=0)
clf_3.fit(count_train, y_train)
pred=clf_3.predict(count_test)
score=metrics.accuracy_score(y_test, pred)
print("accuracy: %0.3f" %score)

cm=metrics.confusion_matrix(y_test, pred, labels=['fake','real'])
```

# Decision Tree

```python
clf_4=DecisionTreeClassifier(criterion = 'entropy', random_state = 0)
clf_4.fit(count_train, y_train)
pred=clf_4.predict(count_test)
score=metrics.accuracy_score(y_test, pred)
print("accuracy: %0.3f" %score)

cm=metrics.confusion_matrix(y_test, pred, labels=['fake','real'])
```

# SVM

```python
clf_5=SVC(kernel = 'rbf', random_state = 0)
clf_5.fit(count_train, y_train)
pred=clf_5.predict(count_test)
score=metrics.accuracy_score(y_test, pred)
print("accuracy: %0.3f" %score)

cm=metrics.confusion_matrix(y_test, pred, labels=['fake','real'])
```
