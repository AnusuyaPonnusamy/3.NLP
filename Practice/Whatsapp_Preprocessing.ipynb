```python
def sentimentalAnalysis(data, columnname):
    import matplotlib.pyplot as plt
    import seaborn as sns
    "Importing Necessary Package"
    from nltk.sentiment.vader import SentimentIntensityAnalyzer
    sid=SentimentIntensityAnalyzer()

    "Deleting null or empty value"
    data.dropna(inplace=True)

    "Checking for a comment"

    "Creating respective columns"

    data['scores'] = data[columnname].apply(lambda commentText: sid.polarity_scores(commentText))
    data['compound']=data['scores'].apply(lambda score_dict:score_dict['compound'])
    data['Negative']=data['scores'].apply(lambda score_dict:score_dict['neg'])
    data['Positive']=data['scores'].apply(lambda score_dict:score_dict['pos'])
    data['Neutral']=data['scores'].apply(lambda score_dict:score_dict['neu'])

    "Creating final pos or neg using compound score"
    data['comp_score']=data['compound'].apply(lambda c: 'pos' if c>=0 else 'neg')
    plt.clf()
    "Checking how many post and neg"
    posneg=pd.DataFrame(data['comp_score'].value_counts())
    return posneg, data
    
```

```python
pos,data_Senti=sentimentalAnalysis(dataset,columnname='Chat')
```

# TOPIC MODELLING - NMF (Non-Negative Matrix Factorization)
It is an unsupervised machine learning technique used to discover hidden topics and patterns in text data.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(max_df=0.95, min_df=2, stop_words='english')
dtm=tfidf.fit_transform(df["Chat"]) # dtm means Document-Term Matrix - A table containing the importance of words in each document/message.
```

```python
from sklearn.decomposition import NMF #NMF means Non-Negative Matrix Factorization - to find hidden topics in your text.
nmf_model=NMF(n_components=5, random_state=42)
nmf_model.fit(dtm)
```

```python
for index, topic in enumerate (nmf_model.components_):
    results=([tfidf.get_feature_names_out()[i] for i in topic.argsort()[10:]]) # [10:] means take the last 10 items — the 10 highest-scoring words.
    print (results)
```

```python
topic_results=nmf_model.transform(dtm)
```

```python
df["Topic"]=topic_results.argmax(axis=1)
```

```python
df["Topic"].value_counts()
```

# TOPIC MODELLING - LDA (Latent Dirichlet Allocation)
It is an unsupervised machine learning technique used for discovering hidden topics in text data.

```python
from sklearn.decomposition import LatentDirichletAllocation

# Convert whatsapp chat text into word-count matrix
tfidf=TfidfVectorizer(max_df=0.95, min_df=2, stop_words='english')
dtm=tfidf.fit_transform(df["Chat"]) 

# LDA model
lda_model = LatentDirichletAllocation(n_components=5, random_state=42)

# Fit LDA model
lda_model.fit(dtm)

# Get all words
words = tfidf.get_feature_names_out()

# Display top 10 words for each topic
for index, topic in enumerate(lda_model.components_):
    top_words = [words[i] for i in topic.argsort()[-10:]]
    print("Topic", index + 1, ":", top_words)
```
